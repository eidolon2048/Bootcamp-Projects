```python
cities = ["Brandenburg", "Hamburg", "Manchester", "Heathrow", "Barcelona"]
weather_dict = {'city': [],
                'forecast_time': [],
                'outlook': [],
                'temperature': [],
                'temperature_feels_like': [],
                'wind_speed': [],
                'wind_deg': [],
                'humidity': [],
                'pressure': []}
```

```python
API_key = "273b832964999f1a9e88c052e7e6528d"
```

```python
for city in cities:
  forecast = requests.get(f"http://api.openweathermap.org/data/2.5/forecast?q={city}&appid={API_key}&units=metric").json()
  for weather in forecast.get("list"):
    weather_dict["city"].append(city)
    weather_dict["forecast_time"].append(weather.get("dt_txt"))
    weather_dict["outlook"].append(weather["weather"][0]["main"])
    weather_dict["temperature"].append(weather.get("main").get("temp"))
    weather_dict["temperature_feels_like"].append(weather.get("main").get("feels_like")
    weather_dict["wind_speed"].append(weather.get("wind").get("speed"))
    weather_dict["wind_deg"].append(weather.get("wind").get("deg"))
    weather_dict["humidity"].append(weather.get("main").get("humidity"))
    weather_dict["pressure"].append(weather.get("main").get("pressure"))
```

```python
weather_df = pd.DataFrame(weather_dict)

weather_df.reset_index(inplace=True)

weather_df.forecast_time = pd.to_datetime(weather_df.forecast_time)

weather_df.rename(columns={"index": "city_id"})
```
